# Simple Linear Regression

Simple Linear Regression is one of the most basic and widely used techniques in Machine Learning and Statistics to model the relationship between two variables.

It tries to find a straight-line relationship between:

● Independent variable (X) → input (e.g., study hours)y = mx + b

● Dependent variable (Y) → output (e.g., exam score)

**y = mx + b**

**y** = predicted value

**x** = input variable

**m**= slope (how much y changes when x changes)

**b** = intercept (value of y when x = 0)

### Example;
If you study more hours (X), your marks (Y) may increase.
The model draws a best-fit line through the data points to predict marks.

### Goal

To minimize the difference between:

● Actual values

● Predicted values

### Applications

● Sales prediction

● Salary estimation

● Demand forecasting

● Trend analysis

## Importing the libraries

## Importing the dataset

```python
dataset=pd.read_csv('Salary_Data.csv')
```

```python
dataset.head()
```

```python
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

Your code is a very common way to split data into features (X) and target (y) using pandas:

: → all rows

:-1 → all columns except the last one(exclude salary column)

-1 → only the last column(salary)

## Splitting the dataset into the Training set and Test set

raining set → to teach the model

Test set → to check if it learned correctly

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=1/3,random_state=0)
```

## Training the Simple Linear Regression Model on the Training set

LinearRegression() → creates the model

fit(x_train, y_train) → trains the model

👉 The model finds the best-fit line:

Learns slope (m)

Learns intercept (b)

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(x_train,y_train)
```

## Predicting the Test Set Results

```python
y_pred=regressor.predict(x_test)
```

## Visualising the training set results

```python
plt.scatter(x_train,y_train,color='red')
plt.plot(x_train,regressor.predict(x_train),color='blue')
plt.title("Salary vs Experience(Training Set)")
plt.xlabel('Year of Experience')
plt.ylabel("salary")
plt.show()
```
